```python
import numpy as np
import pandas as pd

data_path = "test_data.txt"

namespace = {"np": np}
with open(data_path, "r") as f:
    exec(f.read(), namespace)

uniaxial_data = namespace["uniaxial_data"]
shear_data = namespace["shear_data"]

lam_u, P_u = uniaxial_data[:, 0], uniaxial_data[0:, 1]

lam_s, P_s = shear_data[0:, 0], shear_data[0:, 1]
print(lam_s)
print(P_s)


df = pd.DataFrame({
    "stretch_uniaxial": lam_u,
    "stress_uniaxial": P_u,
    "stretch_shear": lam_s,
    "stress_shear": P_s,
})

df.to_csv("combined_data.csv", index=False)
```
