## Importing SymPy
---

The most common import pattern for interactive use:


````python
from sympy import *
from sympy import init_printing, symbols, Eq, Matrix
from IPython.display import display, Markdown

# Define common symbolic variables
x, y, n = symbols('x y n')
a1 = symbols('a_1', integer=True)
a2 = symbols('a_2', positive=True)

print(a1)
display(a1)
display(a1**2)



# Automatically selects the best renderer (MathJax/LaTeX in Jupyter, Unicode in terminal)
init_printing(use_unicode=True)

x, y = symbols('x y')

# Equations automatically render nicely:
pprint(Eq(3*x + 2*y, 12))
Eq(3*x + 2*y, 12)

````

# Basic Commands
---

| Function | Purpose | Example |
|---|---|---|
| `simplify(expr)` | Algebraic simplification | `simplify((x**2 - 4)/(x - 2))` → `x + 2` |
| `expand(expr)` | Expand products and powers | `expand((x + y)**3)` |
| `factor(expr)` | Factor into irreducibles | `factor(x**2 - 5*x + 6)` → `(x-2)(x-3)` |
| `expr.subs(x, val)` | Substitute a value | `expr.subs(x, 5)` |
| `expr.evalf(subs={...})` | Numerical evaluation | `expr.evalf(subs={x: 3})` |
| `apart(expr)` | Partial fraction decomposition | `apart(1/(x**2 + 3*x + 2))` |
| `together(expr)` | Combine fractions | `together(1/(x+1) - 1/(x+2))` |

````python
display(Markdown("**simplify command**"))

display(simplify((x**2 - 4) / (x - 2)))
display(Markdown("**expand command**"))
display(expand((x + y )**3))

display(Markdown("**factor command**"))
print("factor command")
display(factor(x**2 - 5*x + 6))

expr = x**2 + 3*x + 2
display(expr)

print(expr.subs(x, 5))
print(expr)
print(expr.evalf(subs={x: 3}))

# or float(expr.subs(x, 3))

numer(expr)
denom(expr)

display(apart(1 / (x**2 + 3*x + 2)))
display(together(1 / (x + 1) - 1 / (x + 2)))
together(1 / (x + 1) - 1 / (x + 2))
````

# Calculus & Equations 
---


### 1. Differentiation & Integration






````python

# Derivatives
diff(x**3, x)
diff(x * y**2, x, y)

# Integrals
integrate(x**2, x)
integrate(x**2, (x, 0, 1))
````

### 2. Limits & Series





````python


# Limits
limit(sin(x) / x, x, 0)
limit(1 / x, x, oo)

# Series & Summation
display(sin(x).series(x, 0, 5))
summation(1 / n**2, (n, 1, oo))



````

### 3. Solving Equations

````python
# Solving Equations
display(solve(x**2 - 4, x))
solve([x + y - 3, x - y - 1], [x, y])
````

````python
# Differential Equations
f = Function('f')
dsolve(Derivative(f(x), x) + f(x), f(x)) # ODE: f' + f = 0
````

# Linear Algebra
---

````python
A = Matrix([[1, 2], [3, 4]])

A.T
A**2
A.trace()
A.row(0)
A.col(0)
A.shape

A.rank()
A.norm()

A.eigenvals()
A.eigenvects()

B = Matrix([[0, -1], [1, 0]])

v1 = Matrix([1, 2, 3])
v2 = Matrix([4, 5, 6])
v1.dot(v2)
v1.norm()

b = Matrix([7, 10])
linsolve((A, b), [x, y])
A.gauss_jordan_solve(b)
````
